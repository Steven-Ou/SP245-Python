```python
cars = ["Crosstrek", "Forester", "RAV4", "Tuscon", "BRZ", "Supra", "Lucid"]
```

```python
print(cars[0])
```

The reason we rename the `pyplot` module to `plt` is so that we do not have to type `pyplot` repeatedly. This is often the convention or standard when using this library. The same can be said about NumPy as np and matplotlib as mpl.

The `pyplot` module, which is now known as `plt`, contains a number of functions that generate charts and plots. We first start off with a simple line graph.

```python
squares = [1, 4, 9, 16, 25] #This creates a list of data that we will plot
fig, ax = plt.subplots()    #This creates a figure with a set of axes. Note that this uses the plt.subplot() function which has no arguments. If there are no arguments, then this will create a single plot by default.
ax.plot(squares)            #This tells Python on the axes created, to plot the data in our list. Since ONLY ONE list is provided, Python assumes that these are the y-values and the x-values are taken to be indicies of each entry.

plt.show()                  #This tells Python to render the plot on our screen.
```

What if we wanted to plot $f(x) = x^2$ which passes through the points $(1,1), (2,4), (3,9), (4,16), (5,25)$? What we can do is enter two lists -- one list which represents our $x-$values and another list which represents our $y-$values as shown below:

```python
inputs = [1, 2, 3, 4, 5]      #This creates a list of x-inputs
outputs = [1, 4, 9, 16, 25]   #This creates a list of y-inputs

fig, ax = plt.subplots()    #plt.subplots() is a function that returns a figure and axes objects. When the function which has no arguments,  this will create a single plot by default. We will give an example of when you should change the inputs later.
ax.plot(inputs,outputs)    #This tells Python on the axes created, to plot the data in our list. Since TWO lists are provided, Python assumes the first list are the x values and the second list are the corresponding y values.

plt.show()
```

Notice that the default plot is a line graph. As a result, there are a couple of things that are wrong with this. For instance, between any two points, we have a linear line that connects them and we know that quadratics do not behave this way. We will talk about a quick fix in a moment. If, however, we wish to simply plot these points without connecting them, then we use a scatterplot.

To create a scatterplot, we use almost the same code as above. The only difference is that instead of writing `ax.plot(inputs,outputs)` in line (4), we will write `ax.scatter(inputs,outputs)`.

```python
inputs = [1, 2, 3, 4, 5]      #This creates a list of x-inputs
outputs = [1, 4, 9, 16, 25]   #This creates a list of y-inputs

fig, ax = plt.subplots()    #This creates a figure with a set of axes that will be specified in the next line. Note that this uses the plt.subplot() function which has no arguments. If there are no arguments, then this will create a single plot by default.
ax.scatter(inputs,outputs)    #This tells Python on the axes created, to plot the data in our list. Since TWO lists are provided, Python assumes the first list are the x values and the second list are the corresponding y values.

plt.show()
```

The above code answers our question of how to create a scatterplot.

Still, the problem remains of how do we "connect the dots" while preserving the shape of a quadratic?



Answer:

Here is how we would use 5000 points to sketch the graph of $x^2$ on $[0,5]$. We first think of a loop that will generate many points on the interval [0,5]:

```python
x_values = []
for i in range(0,5001,1):
  x_values.append(i/1000)
print(x_values)
```

We placed these values into a list that we called `x_values`.

```python
y_values = []
for i in range(0,5001,1):
  y_values.append((i/1000)**2)
print(y_values)
```

We placed these values into a list that we called `y_values`.

Finally, we plot the lists `x_values` and `y_values`.

```python
inputs = x_values     #This creates a list of x-inputs or we can leave this as x_values
outputs = y_values   #This creates a list of y-inputs or we can leave this as y_values

fig, ax = plt.subplots()    #This creates a figure with a set of axes that will be specified in the next line. Note that this uses the plt.subplot() function which has no arguments. If there are no arguments, then this will create a single plot by default.
ax.plot(inputs,outputs)    #This tells Python on the axes created, to plot the data in our list. Since TWO lists are provided, Python assumes the first list are the x values and the second list are the corresponding y values.

plt.show()
```

Finally, we will discuss how to plot a recursive sequence.

In the loan example discussed in lecture (forthcoming), we derived the following formula:

\begin{align*}
P_i = P_{i-1}(1+r) - M
\end{align*}

where $P_i$ is the outstanding principal after month $i$, $r$ is the monthly interest rate, and $M$ is the monthly payment. In our example, $P_0 =\$ 578850,  M = \$3861.216387, r = 0.005855$, and $N = 360$.

```python
principal_amounts = [578850]                          #create a list the first element being the value of the loan.

for i in range(1,361):                                  #execute the following code 100 times
    principal_amounts.append(principal_amounts[-1]*(1+0.005855)-3861.216387) #This calculates the next term of the sequence and appends the term to our list.
fig, ax = plt.subplots()                             #This creates a figure with a set of axes. Note that this uses the plt.subplot() function which has no arguments. If there are no arguments, then this will create a single plot by default.
ax.plot(principal_amounts, color = 'r', linestyle = 'dashed')                            #This tells Python on the axes created, to plot the data in our list. Since ONLY ONE list is provided, Python assumes that these are the y-values and the x-values are taken to be indicies of each entry.

plt.show()                  #This tells Python to render the plot on our screen.
```
